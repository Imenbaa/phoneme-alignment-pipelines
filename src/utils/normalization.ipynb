```python
import unicodedata

REF_MAPPING = {
    # Consonants
    "Z": "ʒ",
    "S": "ʃ",
    "R": "ʁ",
    "N": "ŋ",
    "J": "ɲ",
    "H": "ɥ",
    "g": "ɡ",
    "Z=": "ʒ",

    # Vowels
    "E": "ɛ",
    "O": "ɔ",
    "2": "ø",
    "9": "œ",
    "@": "ə",

    # Nasals (SAMPA)
    "a~": "ɑ̃",
    "o~": "ɔ̃",
    "e~": "ɛ̃",
    "9~": "ɛ̃",
    "E": "ɛ",
    "m=": "m",
    "n=": "n",
    "9~": "ɛ̃",
}
def normalize_reference(ref_string):
    ref_string=unicodedata.normalize("NFC", ref_string)
    tokens = ref_string.strip().split()

    normalized_tokens = []

    for tok in tokens:

        # skip noise
        if tok in ["_", "sil", "spn", "%", "?", "0", "=","fe~","Ra~","sjo~","~e"]:
            continue

        # map if exists
        tok = REF_MAPPING.get(tok, tok)

        normalized_tokens.append(tok)

    # return space-separated phoneme string
    return normalized_tokens
def normalize_hypothesis(hyp_string):
    hyp_string=unicodedata.normalize("NFC", hyp_string)
    tokens = hyp_string.strip().split()

    normalized_tokens = []

    for tok in tokens:

        # --- multilingual projection ---
        tok = HYP_PROJECTION.get(tok, tok)

        # --- nasal canonicalization ---
        tok = NASAL_CANONICAL.get(tok, tok)

        # --- remove suprasegmentals ---
        tok = tok.replace("ː", "")
        tok = tok.replace('̃', "")

        if tok != "":
            normalized_tokens.append(tok)

    return normalized_tokens
```
