```python
import os
import csv
import textgrid  # pip install praat-textgrids or textgrid
from pathlib import Path

def extract_phonemes_from_textgrid(tg_path, tier_name="phones"):
    """
    Extract phoneme sequence from a TextGrid file.
    Returns a space-separated string of phonemes.
    """
    tg = textgrid.TextGrid.fromFile(str(tg_path))

    # Find the correct tier
    tier = None
    for t in tg.tiers:
        if t.name.lower() == tier_name.lower():
            tier = t
            break

    if tier is None:
        raise ValueError(f"Tier '{tier_name}' not found in {tg_path}")

    phonemes = []
    for interval in tier.intervals:
        label = interval.mark.strip()
        if label != "" and label.lower() != "sil":
            phonemes.append(label)

    return " ".join(phonemes)


def generate_csv_from_textgrids(textgrid_dir, output_csv, tier_name="phones"):
    """
    Generate CSV with:
        audio_filename | reference_phonemes
    """
    textgrid_dir = Path(textgrid_dir)

    rows = []

    for tg_file in textgrid_dir.glob("*.TextGrid"):
        try:
            phoneme_seq = extract_phonemes_from_textgrid(tg_file, tier_name)

            audio_filename = tg_file.stem + ".wav"

            rows.append({
                "audio_filename": audio_filename,
                "reference_phonemes": phoneme_seq
            })

        except Exception as e:
            print(f"Error processing {tg_file}: {e}")

    # Write CSV
    with open(output_csv, mode="w", newline="", encoding="utf-8") as f:
        writer = csv.DictWriter(f, fieldnames=["audio_filename", "reference_phonemes"])
        writer.writeheader()
        writer.writerows(rows)

    print(f"CSV saved to {output_csv}")
```
